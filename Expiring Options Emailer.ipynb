```python
import smtplib
from email.mime.text import MIMEText

def send_sms_via_email(phone_number, carrier, message):
    # Define the email-to-SMS gateway
    carriers = {
        'att':    'txt.att.net',
        'tmobile': 'tmomail.net',
        'verizon':  'vtext.com',
        'sprint': 'messaging.sprintpcs.com'
    }
    
    if carrier not in carriers:
        raise ValueError("Carrier not supported")

    to_number = f'{phone_number}@{carriers[carrier]}'
    auth = ('dougsack@gmail.com', "mojf zvyq rvqk rkxl")

    # Create the email content
    msg = MIMEText(message)
    msg['From'] = auth[0]
    msg['To'] = to_number
    msg['Subject'] = 'SMS via Email'

    try:
        # Send the email via SMTP
        server = smtplib.SMTP("smtp.gmail.com", 587)
        server.starttls()
        server.login(auth[0], auth[1])
        server.sendmail(auth[0], to_number, msg.as_string())
        server.quit()
        print("SMS sent successfully")
    except Exception as e:
        print(f"Failed to send SMS: {e}")

```

```python
from datetime import datetime

doc_list = []

for _, row in final_data.iterrows():
    #changes_val = row['changes']
    #changes_val = str(changes_val) if pd.notna(changes_val) else ""

    new_update = {
        'update_date': row['update_date'],
        'Status': row['Status'],
        'title': row['Title'],
        'ITD_2023': row['Inception To Date | 2023'],
        'ITD_2024': row['Inception To Date | 2024'],
        'ITD_2025': row['ITD TOTAL SPEND | 2025'],
        'ITD_2026': row['ITD TOTAL SPEND | 2026'],
        '2024_spend': row['Year To Date | 2024'],
        '2025_spend': row['Year To Date | 2025'],
        '2026_spend': row['Year To Date | 2026'],
        'Remaining_Committed': row['Committed '],
        'Remaining_Optional': row['Optional'],
        'Total_Cost_with_Committments': row['ITD w Commitments | 2026'],
        'Changes':row['Changes from Prior'],
        "Summary":row['SUMMARY']
    }

    doc = {
        '_id': row['_id'],
        #'date_added':row['DATE ADDED'],
        'title': row['Title'],
        'division': row['division'].lower(),
        'Status': row['Status'],
        'update_date': row['update_date'],
        'Total_Cost_with_Committments': new_update['Total_Cost_with_Committments'],
        'ITD_2026': new_update['ITD_2026'],
        'Remaining_Committed': new_update['Remaining_Committed'],
        'Remaining_Optional': new_update['Remaining_Optional'],
        'Changes': new_update['Changes'],
        # This will *append* to the history on upsert
        '$push': {'updates': new_update}
    }

    doc_list.append(doc)
```

```python
from pymongo import UpdateOne

operations = []

for doc in doc_list:
    # Split update payloads
    base_fields = {
        'title': doc['title'],
        'division': doc['division'],
        'status':doc['Status'],
        'update_date': doc['update_date'],
        'Total_Cost_with_Committments': doc['Total_Cost_with_Committments'],
        'ITD_current_year': doc['ITD_2026'],
        'Remaining_Committed': doc['Remaining_Committed'],
        'Remaining_Optional': doc['Remaining_Optional'],
    }

    new_update = doc['$push']['updates']  # from your earlier construction

    operations.append(
        UpdateOne(
            {"_id": doc["_id"]},
            {
                "$set": base_fields,
                "$push": {"updates": new_update}
            },
            upsert=True
        )
    )

result = collection.bulk_write(operations)
print(f"Matched: {result.matched_count}, Upserted: {len(result.upserted_ids)}, Modified: {result.modified_count}")
```
